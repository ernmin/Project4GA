```python
database_schema = {
    'user_course_info_table': df_user_course_info,
    'teacher_stats_table': df_teacher_stats,
    'pro_course_table': df_pro_course,
    'price_list_table': df_price_list,
    'other_languages_table': df_also_speaks,
    'other_languages_ref_table': df_also_speaks_reference
}
```

```python
conn = sqlite3.connect('italki_analytics.db')

for table_name, dataframe in database_schema.items():
    
    # Check if the index has an explicit name assigned
    if dataframe.index.name is not None:
        pk_label = dataframe.index.name
    else:
        pk_label = 'id' 
    
    # Write cleanly to SQLite
    dataframe.to_sql(
        name=table_name, 
        con=conn, 
        if_exists='replace', 
        index=True, 
        index_label=pk_label
    )
    print(f"Successfully loaded '{table_name}' using primary key: {pk_label}")

# Always close the connection for DB Browser access
conn.close()
```
