```python
def json_existing(first_page, dictionary):
    with open("english_price.json", 'r') as file:
            # Load existing data into a dictionary
            file_data = json.load(file)
            
    if str(first_page) in file_data:
          print('Error: Duplicating Page')
          return False
    else:
        file_data[str(first_page)] = dictionary        
        with open("english_price.json", 'w') as file:
                # Load existing data into a dictionary
                json.dump(file_data, file, indent=4)
```
