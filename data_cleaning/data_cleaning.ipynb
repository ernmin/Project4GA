Check for any duplicates in the also speaks file\
Check if there are teachers in the 'also_speaks' file but missing in the 'user_course_info' file.

```python
with open('../data_extraction/also_speaks.json', 'r') as file:
    also_speaks_check = json.load(file)

df_also_speaks = pd.DataFrame(also_speaks_check)
```

No exact duplicates found in the also speaks dataframe.\
Exact duplicates need to be checked because with the user_course_info table, there is a one to many relationship.\
There are teachers who speak multiple other languages so they will have multiple entries in the 'also speaks' table

```python
all_duplicates = df_also_speaks[df_also_speaks.duplicated(keep=False)]
all_duplicates
```

```python
with open('./user_course_info_cleaned.json', 'r') as file:
    user_course_info = json.load(file)

df_user_course_info = pd.DataFrame(user_course_info)
```

```python
df_num_of_languages = df_also_speaks.groupby('user_id').count().reset_index()
```

```python
df_num_of_languages[df_num_of_languages['language'] > 0]
```

```python
df_merged = pd.merge(df_user_course_info, df_num_of_languages, on='user_id', how='left_anti')
```
